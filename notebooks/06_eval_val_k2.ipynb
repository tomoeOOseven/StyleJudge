```python
# ╔══════════════════════════════════════════════════════════════════════════════╗
# ║  ELITE LORA SFT — Qwen3.5-0.8B · Unsloth · Visual Style Judge         ║
# ║  Platform : Kaggle Offline Notebook                                        ║
# ║  Task     : Score artistic style-transfer quality (3-metric XML output)   ║
# ╚══════════════════════════════════════════════════════════════════════════════╝


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 6 ── MODEL LOADING  (Unsloth FastVisionModel — Full bf16, No QLoRA)
# ═══════════════════════════════════════════════════════════════════════════════
import os
import sys

# 1. CRITICAL: Inject your offline library path FIRST
LOCAL_LIB = "/kaggle/working/lib"
if LOCAL_LIB not in sys.path:
    sys.path.insert(0, LOCAL_LIB)
os.environ["LD_LIBRARY_PATH"] = "/usr/local/cuda/lib64:" + os.environ.get("LD_LIBRARY_PATH", "")
from unsloth import FastVisionModel

print("[MODEL] Loading Qwen3.5-0.8B-Instruct …")
flush()

model, tokenizer = FastVisionModel.from_pretrained(
    model_name      = MODEL_PATH,
    max_seq_length  = MAX_SEQ_LEN,
    dtype           = torch.bfloat16,   # Native Blackwell precision
    load_in_4bit    = False,            # Full precision — we have 95 GB VRAM
    local_files_only= True,             # Kaggle offline
    trust_remote_code = True,
)

print(f"[MODEL] Loaded. Parameters: {sum(p.numel() for p in model.parameters())/1e6:.1f}M")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 8 ── SCORE EXTRACTION UTILITY  (Matches CSV Logic EXACTLY)
# ═══════════════════════════════════════════════════════════════════════════════
import re
from typing import Dict

def extract_multi_metrics(llm_output: str) -> Dict[str, any]:
    """
    Parse 3 integer scores from XML output and compute the final scaled 1-10 score.
    """
    raw = {"sf": None, "cp": None, "rq": None}
    
    # Extract using regex
    sf_m = re.search(r'Style Fidelity:\s*([0-9]+)', llm_output, re.I)
    cp_m = re.search(r'Content Preservation:\s*([0-9]+)', llm_output, re.I)
    rq_m = re.search(r'Rendering Quality:\s*([0-9]+)', llm_output, re.I)
    
    raw["sf"] = max(1, min(10, int(sf_m.group(1)))) if sf_m else 0
    raw["cp"] = max(1, min(10, int(cp_m.group(1)))) if cp_m else 0
    raw["rq"] = max(1, min(10, int(rq_m.group(1)))) if rq_m else 0

    # Compute final rounded integer matching the inference script
    final_scaled_score = min(raw["rq"], raw["cp"]) if (raw["rq"] <= 3 or raw["cp"] <= 3) else max(1, min(10, raw["sf"]))

    return {
        "sf": raw["sf"], 
        "cp": raw["cp"], 
        "rq": raw["rq"], 
        "final_scaled_score": final_scaled_score,
        "raw_extraction": raw
    }
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# CELL 13 ── POST-TRAINING VALIDATION PREDICTIONS  (manual score extraction)
# ═══════════════════════════════════════════════════════════════════════════════

import os
import json
import random
import statistics
import gc
import torch
from safetensors.torch import load_file, save_file
from qwen_vl_utils import process_vision_info

FastVisionModel.for_inference(model)   # re-enable inference-mode KV-cache

# Fix & load adapter
original_adapter_dir = "/kaggle/input/models/sachitbansal2006/qwen3-5-0-8b-10kbalanced-k2/transformers/default/1/adapters_qwen3.5_0.8b"
fixed_adapter_dir    = "/kaggle/working/fixed_lora_adapter"
os.makedirs(fixed_adapter_dir, exist_ok=True)

# Remap keys
original_safetensors = os.path.join(original_adapter_dir, "adapter_model.safetensors")
fixed_safetensors    = os.path.join(fixed_adapter_dir,    "adapter_model.safetensors")

print(f"[INFER] Remapping adapter keys...")
state_dict = load_file(original_safetensors)
new_state_dict = {
    key.replace("model.layers.", "model.language_model.layers."): value
    for key, value in state_dict.items()
}
save_file(new_state_dict, fixed_safetensors)

# Patch adapter_config.json to match the actual saved rank (32)
with open(os.path.join(original_adapter_dir, "adapter_config.json"), "r") as f:
    adapter_config = json.load(f)

adapter_config["r"] = 16
adapter_config["lora_alpha"] = 32  # matches LORA_ALPHA from training

with open(os.path.join(fixed_adapter_dir, "adapter_config.json"), "w") as f:
    json.dump(adapter_config, f, indent=2)

print(f"[INFER] Loading fixed adapter...")
model.load_adapter(fixed_adapter_dir, adapter_name="default")


val_predictions = []
# Ensure seed_value and num_evals are defined previously in your notebook
random.seed(seed_value) 

# Generate 'num_evals' unique random indices from the total length of the dataframe
from sklearn.model_selection import train_test_split
try:
    _, sample_df = train_test_split(
        val_ds.df, 
        test_size=num_evals, 
        stratify=val_ds.df['final_score'], 
        random_state=seed_value
    )
    random_indices = sample_df.index.tolist()
except Exception as e:
    print(f"Stratification failed ({e}), falling back to random sampling.")
    random_indices = random.sample(range(len(val_ds.df)), num_evals)

print(f"[INFER] Running predictions on {num_evals} val samples …")

# FIX 3: Use enumerate to track the loop iteration step properly
for step, i in enumerate(random_indices):
    
    # Fetch the row from the DataFrame using the random index
    row = val_ds.df.iloc[i]

    # FIX 1: Swapped 'sample' for 'row' everywhere
    style_desc = STYLE_DESCRIPTIONS.get(row["style"], "distinctive textures, colors, and brushstrokes.")

    user_content = build_user_content(
        load_image(row["ex_a_src_path"]), load_image(row["ex_a_dst_path"]),
        load_image(row["ex_b_src_path"]), load_image(row["ex_b_dst_path"]),
        load_image(row["q_src_path"]),    load_image(row["q_res_path"]),
        markers=style_desc
    )

    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": user_content},
    ]

    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    image_inputs, _ = process_vision_info(messages)

    inputs = tokenizer(
        text=[text], images=image_inputs,
        padding=True, return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        out_ids = model.generate(
            **inputs,
            max_new_tokens    = 2048,
            do_sample         = False,
            eos_token_id      = tokenizer.eos_token_id,
            pad_token_id      = tokenizer.pad_token_id,
        )

    raw_out = tokenizer.batch_decode(
        out_ids[:, inputs.input_ids.shape[1]:],
        skip_special_tokens=True
    )[0]

    predicted = extract_multi_metrics(raw_out)
    
    # FIX 1: Swapped 'sample' for 'row'
    gt_score = int(row["final_score"])
    final_match = (predicted["final_scaled_score"] == gt_score)

    val_predictions.append({
        "sample_idx":       i,
        "style":            row["style"],       # FIX 1
        "morph_idx":        row["morph_idx"],   # FIX 1
        "gt_final_score":   gt_score,
        
        "pred_sf":          predicted["sf"],
        "pred_cp":          predicted["cp"],
        "pred_rq":          predicted["rq"],
        "pred_final_score": predicted["final_scaled_score"],
        
        "final_score_match": final_match,
        "raw_model_output": raw_out,
        "output_length":    len(raw_out),
    })

    # FIX 2: Replaced flush() with actual garbage collection
    del inputs, out_ids, image_inputs
    gc.collect()
    torch.cuda.empty_cache()

    # FIX 3: Use 'step' for running accuracy and modulus logic
    if (step + 1) % 10 == 0:
        matches = sum(1 for p in val_predictions if p["final_score_match"])
        print(f"  [{step+1:>3}/{num_evals}] style={val_predictions[-1]['style']:<14} | "
              f"Pred: {val_predictions[-1]['pred_final_score']} | "
              f"GT: {val_predictions[-1]['gt_final_score']} | "
              f"Running Acc: {matches}/{step+1}")

# ── Print summary stats ───────────────────────────────────────────────────────
if val_predictions:
    
    # FIX 4: Use a dynamic denominator instead of hardcoded 50
    total_evals = len(val_predictions)
    
    score_errors = [abs(p["pred_final_score"] - p["gt_final_score"]) for p in val_predictions]
    match_count = sum(1 for p in val_predictions if p["final_score_match"])
    close_match_count = sum(1 for p in val_predictions if abs(p["pred_final_score"] - p["gt_final_score"]) <= 1)
    
    print("\n" + "═"*70)
    print(f" Val Prediction Summary ({total_evals} samples)")
    print("═"*70)
    print(f"  Final Score MAE    : {statistics.mean(score_errors):.3f} points off on average")
    print(f"  Exact Matches      : {match_count}/{total_evals} ({100*match_count/total_evals:.1f}%)")
    print(f"  Within +/- 1 Point : {close_match_count}/{total_evals} ({100*close_match_count/total_evals:.1f}%)")
    print(f"  Avg output len     : {statistics.mean(p['output_length'] for p in val_predictions):.0f} characters")
    print("═"*70)
```
