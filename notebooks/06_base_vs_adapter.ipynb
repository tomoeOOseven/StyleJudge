```python
import pandas as pd

# Drop rows where any of these specific columns are empty
df_sft = df_sft.dropna(subset=['rq', 'cp', 'sf'])
# Define a function that takes a row (Series) as input
def calculate_holistic_score(row):
    """Composite score S_final (Eq. 1): min(RQ, CP) if RQ <= 3 or CP <= 3, else SF."""
    if pd.isna(row['rq']) or pd.isna(row['cp']) or pd.isna(row['sf']):
        return 1  # default for broken data
    rq, cp, sf = row['rq'], row['cp'], row['sf']
    if rq <= 3 or cp <= 3:
        return int(min(rq, cp))
    return int(sf)

# Apply the function across rows (axis=1) and update the column
df_sft['final_score'] = df_sft.apply(calculate_holistic_score, axis=1)

# Save back to CSV
df_sft.to_csv('style_rationale_harmonic_110part2.csv', index=False)
```

```python
# ╔══════════════════════════════════════════════════════════════════════════════╗
# ║  ELITE LORA SFT — Qwen2.5-VL-0.8B · Unsloth · Visual Style Judge         ║
# ║  Platform : Kaggle Offline Notebook                                        ║
# ║  Task     : Score artistic style-transfer quality (3-metric XML output)   ║
# ╚══════════════════════════════════════════════════════════════════════════════╝


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 6 ── MODEL LOADING  (Unsloth FastVisionModel — Full bf16, No QLoRA)
# ═══════════════════════════════════════════════════════════════════════════════
import os
import sys

# 1. CRITICAL: Inject your offline library path FIRST
LOCAL_LIB = "/kaggle/working/lib"
if LOCAL_LIB not in sys.path:
    sys.path.insert(0, LOCAL_LIB)
os.environ["LD_LIBRARY_PATH"] = "/usr/local/cuda/lib64:" + os.environ.get("LD_LIBRARY_PATH", "")
from unsloth import FastVisionModel

print("[MODEL] Loading Qwen2.5-VL-0.8B-Instruct …")
flush()

model, tokenizer = FastVisionModel.from_pretrained(
    model_name      = MODEL_PATH,
    max_seq_length  = MAX_SEQ_LEN,
    dtype           = torch.bfloat16,   
    load_in_4bit    = False,            
    local_files_only= True,             # Kaggle offline
    trust_remote_code = True,
)

print(f"[MODEL] Loaded. Parameters: {sum(p.numel() for p in model.parameters())/1e6:.1f}M")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 8 ── SCORE EXTRACTION UTILITY  (Matches CSV Logic EXACTLY)
# ═══════════════════════════════════════════════════════════════════════════════
import re
from typing import Dict

def extract_multi_metrics(llm_output: str) -> Dict[str, any]:
    """
    Parse 3 integer scores from XML output and compute the final scaled 1-10 score,
    matching the exact mangled/artifacted penalty logic used to build the CSV.
    """
    raw = {"sf": None, "cp": None, "rq": None}
    
    # Extract using regex
    sf_m = re.search(r'Style Fidelity:\s*([0-9]+)', llm_output, re.I)
    cp_m = re.search(r'Content Preservation:\s*([0-9]+)', llm_output, re.I)
    rq_m = re.search(r'Rendering Quality:\s*([0-9]+)', llm_output, re.I)
    
    raw["sf"] = max(1, min(10, int(sf_m.group(1)))) if sf_m else 5
    raw["cp"] = max(1, min(10, int(cp_m.group(1)))) if cp_m else 5
    raw["rq"] = max(1, min(10, int(rq_m.group(1)))) if rq_m else 5

    # 🟢 FIXED: Implement the exact penalty logic from your holistic calculation
    weighted_score = (raw["sf"] * 0.40) + (raw["cp"] * 0.30) + (raw["rq"] * 0.30)
    
    if raw["rq"] <= 3 or raw["cp"] <= 3:
        final_scaled_score = min(raw["rq"], raw["cp"])
    else:
        final_scaled_score = max(1, min(10, int(round(weighted_score))))

    return {
        "sf": raw["sf"], 
        "cp": raw["cp"], 
        "rq": raw["rq"], 
        "final_scaled_score": final_scaled_score,
        "raw_extraction": raw
    }
```

```python
# ═══════════════════════════════════════════════════════════════════════════════
# BASE MODEL INFERENCE ── raw Qwen2.5-VL-0.8B, no adapter, no fine-tuning
# ═══════════════════════════════════════════════════════════════════════════════

FastVisionModel.for_inference(model)   # enable inference-mode KV-cache on raw base model

from qwen_vl_utils import process_vision_info
base_val_predictions = []

print(f"[BASE INFER] Running predictions on {min(1000, len(val_ds))} val samples …")

for i in range(min(1000, len(val_ds))):
    sample = val_ds.samples[i]

    style_desc = STYLE_DESCRIPTIONS.get(sample["style"], "distinctive textures, colors, and brushstrokes.")

    user_content = build_user_content(
        load_image(sample["ex_a_src_path"]), load_image(sample["ex_a_dst_path"]),
        load_image(sample["ex_b_src_path"]), load_image(sample["ex_b_dst_path"]),
        load_image(sample["q_src_path"]),    load_image(sample["q_res_path"]),
        sample["style"], style_desc
    )

    messages = [
        {"role": "system", "content": SYSTEM_PROMPT},
        {"role": "user",   "content": user_content},
    ]

    text = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    image_inputs, _ = process_vision_info(messages)

    inputs = tokenizer(
        text=[text], images=image_inputs,
        padding=True, return_tensors="pt"
    ).to(model.device)

    with torch.inference_mode():
        out_ids = model.generate(
            **inputs,
            max_new_tokens    = 1024,  # Safe cap
            do_sample         = False, # Greedy
            eos_token_id      = tokenizer.eos_token_id,
            pad_token_id      = tokenizer.pad_token_id,
        )

    raw_out = tokenizer.batch_decode(
        out_ids[:, inputs.input_ids.shape[1]:],
        skip_special_tokens=True
    )[0]

    predicted = extract_multi_metrics(raw_out)

    gt_score = int(sample["final_score"])
    final_match = (predicted["final_scaled_score"] == gt_score)

    base_val_predictions.append({
        "sample_idx":       i,
        "style":            sample["style"],
        "morph_idx":        sample["morph_idx"],
        "gt_final_score":   gt_score,

        "pred_sf":          predicted["sf"],
        "pred_cp":          predicted["cp"],
        "pred_rq":          predicted["rq"],
        "pred_final_score": predicted["final_scaled_score"],

        "final_score_match": final_match,
        "raw_model_output": raw_out,
        "output_length":    len(raw_out),
    })

    del inputs, out_ids, image_inputs
    flush()

    if (i + 1) % 10 == 0:
        matches = sum(1 for p in base_val_predictions if p["final_score_match"])
        print(f"  [{i+1:>3}/50] style={base_val_predictions[-1]['style']:<14} | "
              f"Pred: {base_val_predictions[-1]['pred_final_score']} | "
              f"GT: {base_val_predictions[-1]['gt_final_score']} | "
              f"Running Acc: {matches}/{i+1}")

# ── Print summary stats ───────────────────────────────────────────────────────────────────────
if base_val_predictions:
    import statistics

    score_errors = [abs(p["pred_final_score"] - p["gt_final_score"]) for p in base_val_predictions]
    match_count = sum(1 for p in base_val_predictions if p["final_score_match"])
    close_match_count = sum(1 for p in base_val_predictions if abs(p["pred_final_score"] - p["gt_final_score"]) <= 1)

    print("\n" + "═"*70)
    print(" Base Model Val Prediction Summary (50 samples)")
    print("═"*70)
    print(f"  Final Score MAE    : {statistics.mean(score_errors):.3f} points off on average")
    print(f"  Exact Matches      : {match_count}/50 ({100*match_count/50:.1f}%)")
    print(f"  Within +/- 1 Point : {close_match_count}/50 ({100*close_match_count/50:.1f}%)")
    print(f"  Avg output len     : {statistics.mean(p['output_length'] for p in base_val_predictions):.0f} characters")
    print("═"*70)
```
