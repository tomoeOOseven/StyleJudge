```python
import os
import shutil
import random

# 1. Define your paths
# Replace these strings with the actual folder names of your Kaggle datasets
dataset_1_path = '/kaggle/input/datasets/chandavandann/global-fish-species-image-dataset-inaturalist'
dataset_2_path = '/kaggle/input/datasets/dimensi0n/lhq-1024'
dataset_3_path = '/kaggle/input/datasets/somanshu7527/more-images-style-transformation-1813'
output_path = '/kaggle/working/combined_dataset/'

# Create the output directory if it doesn't exist
if not os.path.exists(output_path):
    os.makedirs(output_path)

def collect_images(source_dir, sample_size=None):
    """
    Gathers all image files from a directory (including subfolders).
    If sample_size is provided, it returns a random sample.
    """
    valid_extensions = ('.jpg', '.jpeg', '.png', '.bmp', '.tiff')
    all_images = []
    
    for root, dirs, files in os.walk(source_dir):
        for file in files:
            if file.lower().endswith(valid_extensions):
                all_images.append(os.path.join(root, file))
    
    if sample_size and len(all_images) >= sample_size:
        return random.sample(all_images, sample_size)
    return all_images

# 2. Process Datasets
# Sample 550 from Dataset 1
images_1 = collect_images(dataset_1_path, sample_size=550)

# Sample 550 from Dataset 2
images_2 = collect_images(dataset_2_path, sample_size=550)

# Take all from Dataset 3
images_3 = collect_images(dataset_3_path)

# Combine lists
final_image_list = images_1 + images_2 + images_3

# 3. Copy files to the new directory
print(f"Starting copy of {len(final_image_list)} images...")

for i, img_path in enumerate(final_image_list):
    # We add a prefix (i_) to prevent filename collisions if different datasets 
    # have files with the same name (e.g., "001.jpg")
    file_name = f"{i}_{os.path.basename(img_path)}"
    dest_path = os.path.join(output_path, file_name)
    shutil.copy(img_path, dest_path)

print(f"Done! New dataset created at: {output_path}")
print(f"Total images: {len(os.listdir(output_path))}")
```

```python
import json
import os
import shutil
from kaggle_secrets import UserSecretsClient

# 1. Load the two SEPARATE secrets
user_secrets = UserSecretsClient()
os.environ['KAGGLE_USERNAME'] = user_secrets.get_secret("somanshu7527")
os.environ['KAGGLE_KEY'] = user_secrets.get_secret("somanshu7527")

# 2. Setup your paths
# Ensure you have already run the !zip command to create this file
zip_file_path = '/kaggle/working/combined_data.zip' 
staging_dir = '/kaggle/working/upload_staging/'
username = os.environ['KAGGLE_USERNAME']

# We use dashes and lowercase because Kaggle is strict about the dataset ID
dataset_slug = 'dataset-dl-project-2913images' 

# 3. Create the staging area
if os.path.exists(staging_dir):
    shutil.rmtree(staging_dir) # Clear old staging data if it exists
os.makedirs(staging_dir)

# 4. Move the zip file into the staging folder
if os.path.exists(zip_file_path):
    shutil.copy(zip_file_path, os.path.join(staging_dir, 'combined_data.zip'))
else:
    print("Error: Zip file not found! Make sure you zipped your images first.")

# 5. Create the metadata
metadata = {
    "title": "DL Project 2913 Images",
    "id": f"{username}/{dataset_slug}",
    "licenses": [{"name": "CC0-1.0"}]
}

with open(os.path.join(staging_dir, 'dataset-metadata.json'), 'w') as f:
    json.dump(metadata, f, indent=4)

print(f"Staging complete. Ready to upload to: {username}/{dataset_slug}")

# 6. Final Upload
!kaggle datasets create -p {staging_dir}
```
