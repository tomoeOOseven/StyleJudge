```python
# ╔══════════════════════════════════════════════════════════════════════════════╗
# ║  GRPO FINE-TUNING — Qwen2.5-VL-0.8B · Unsloth · Visual Style Judge        ║
# ║  Continues from SFT LoRA adapter (best checkpoint)                         ║
# ║  Hardware : NVIDIA RTX 6000 Pro Blackwell  (95 GB VRAM)                   ║
# ║  Platform : Kaggle Offline Notebook                                        ║
# ║  Algorithm: Group Relative Policy Optimization (DeepSeek-R1 style)        ║
# ╚══════════════════════════════════════════════════════════════════════════════╝

# ═══════════════════════════════════════════════════════════════════════════════
# CELL 14 ── GRPO IMPORTS & HYPER-PARAMETERS
# ═══════════════════════════════════════════════════════════════════════════════
import torch, gc, re, json, os, sys, copy, time, random, math
import numpy as np
import pandas as pd
from pathlib import Path
from datetime import datetime
from typing import List, Dict, Optional, Tuple
from PIL import Image
from torch.utils.data import Dataset, DataLoader

# ── Kaggle offline lib path (same as SFT cells) ──────────────────────────────
LOCAL_LIB = "/kaggle/working/lib"
if LOCAL_LIB not in sys.path:
    sys.path.insert(0, LOCAL_LIB)

from unsloth import FastVisionModel
from qwen_vl_utils import process_vision_info

# ── Paths (reuse SFT outputs) ─────────────────────────────────────────────────
SFT_ADAPTER_PATH  = "/kaggle/input/datasets/sachitbansal2006/qwen3-5-train-adpaters/adapters_qwen3.5_0.8b"      # best SFT checkpoint
GRPO_OUTPUT_DIR   = "/kaggle/working/grpo_adapter"
GRPO_LOG_CSV      = "/kaggle/working/grpo_training_logs.csv"
MODEL_PATH        = "/kaggle/input/models/barnobarno/qwen3.5-0.8b/transformers/unsloth/1"

# ── GRPO Hyper-parameters ─────────────────────────────────────────────────────
GRPO_GROUP_SIZE   = 6       # G: completions sampled per prompt
GRPO_MINI_BATCH   = 2       # prompts per grad step (effective batch = G × mini = 12)
GRPO_GRAD_ACCUM   = 4       # effective update batch = 48 completions
GRPO_LR           = 5e-6    # lower than SFT; policy fine-tuning is fragile
GRPO_EPOCHS       = 1       # 1 GRPO epoch ≈ 3 SFT epochs in refinement power
GRPO_KL_COEF      = 0.04    # β — KL penalty weight vs reference model
GRPO_CLIP_EPS     = 0.2     # PPO-style probability ratio clipping
GRPO_MAX_NEW_TOKENS = 900   # max generation length per completion
GRPO_EVAL_STEPS   = 50
GRPO_LOG_STEPS    = 10
GRPO_WARMUP_STEPS = 20
MAX_SEQ_LEN       = 4096
IMAGE_SIZE        = 1024
SEED              = 42

# ── Reward weights (must sum to 1.0) ─────────────────────────────────────────
W_FORMAT          = 0.15    # XML structure reward
W_SCORE_ACC       = 0.55    # final score accuracy (primary signal)
W_METRIC_ACC      = 0.20    # sub-metric accuracy (SF / CP / RQ)
W_REASONING       = 0.10    # reasoning chain quality reward

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

def flush():
    gc.collect()
    torch.cuda.empty_cache()
    if torch.cuda.is_available():
        torch.cuda.synchronize()

print("[GRPO] Configuration loaded.")
print(f"  Group size G            : {GRPO_GROUP_SIZE}")
print(f"  Effective completion batch: {GRPO_GROUP_SIZE * GRPO_MINI_BATCH * GRPO_GRAD_ACCUM}")
print(f"  KL coefficient β        : {GRPO_KL_COEF}")
print(f"  Score accuracy weight   : {W_SCORE_ACC}")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 15 ── LOAD SFT MODEL AS GRPO POLICY + FROZEN REFERENCE MODEL
# ═══════════════════════════════════════════════════════════════════════════════
flush()

print("[GRPO] Loading base model...")
model, tokenizer = FastVisionModel.from_pretrained(
    model_name        = MODEL_PATH,
    max_seq_length    = MAX_SEQ_LEN,
    dtype             = torch.bfloat16,
    load_in_4bit      = False,
    local_files_only  = True,
    trust_remote_code = True,
)

# ── Load the SFT LoRA adapter onto the policy ─────────────────────────────────
print(f"[GRPO] Loading SFT adapter from {SFT_ADAPTER_PATH} ...")
from peft import PeftModel
model = PeftModel.from_pretrained(model, SFT_ADAPTER_PATH, is_trainable=True)

print(f"[GRPO] Policy model loaded. Trainable params: "
      f"{sum(p.numel() for p in model.parameters() if p.requires_grad)/1e6:.2f}M")

# ── Build frozen reference model (same SFT weights, no grad) ─────────────────
# We keep reference in eval mode to compute KL divergence during GRPO
print("[GRPO] Cloning frozen reference model...")

ref_model, _ = FastVisionModel.from_pretrained(
    model_name        = MODEL_PATH,
    max_seq_length    = MAX_SEQ_LEN,
    dtype             = torch.bfloat16,
    load_in_4bit      = False,
    local_files_only  = True,
    trust_remote_code = True,
)
ref_model = PeftModel.from_pretrained(ref_model, SFT_ADAPTER_PATH, is_trainable=False)
ref_model.eval()

# Freeze ALL reference model params
for p in ref_model.parameters():
    p.requires_grad_(False)

print("[GRPO] Reference model frozen. ✅")
flush()


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 16 ── REWARD FUNCTIONS  (The Heart of GRPO)
# ═══════════════════════════════════════════════════════════════════════════════

# Reuse score extraction from SFT cells
def extract_multi_metrics(llm_output: str) -> Dict:
    sf_m = re.search(r'Style Fidelity:\s*([0-9]+(?:\.[0-9]+)?)', llm_output, re.I)
    cp_m = re.search(r'Content Preservation:\s*([0-9]+(?:\.[0-9]+)?)', llm_output, re.I)
    rq_m = re.search(r'Rendering Quality:\s*([0-9]+(?:\.[0-9]+)?)', llm_output, re.I)

    sf = max(1, min(10, int(round(float(sf_m.group(1)))))) if sf_m else None
    cp = max(1, min(10, int(round(float(cp_m.group(1)))))) if cp_m else None
    rq = max(1, min(10, int(round(float(rq_m.group(1)))))) if rq_m else None

    # Graceful fallbacks
    sf = sf or 5;  cp = cp or 5;  rq = rq or 5

    weighted = (sf * 0.40) + (cp * 0.30) + (rq * 0.30)
    if rq <= 3 or cp <= 3:
        final = min(rq, cp)
    else:
        final = max(1, min(10, int(round(weighted))))

    return {"sf": sf, "cp": cp, "rq": rq, "final": final,
            "sf_found": sf_m is not None, "cp_found": cp_m is not None,
            "rq_found": rq_m is not None}


def reward_format(completion: str) -> float:
    """
    Checks for correct XML skeleton presence.
    Returns 0.0 – 1.0 linearly based on how many required tags are present.
    """
    required_tags = [
        r"<Thinking_Process>",  r"</Thinking_Process>",
        r"<Quantitative_Metrics>", r"</Quantitative_Metrics>",
        r"<Final_Output>",      r"</Final_Output>",
        r"Style Fidelity:",
        r"Content Preservation:",
        r"Rendering Quality:",
        r"Brief Rationale:",
    ]
    hits = sum(1 for tag in required_tags if re.search(tag, completion, re.I))
    return hits / len(required_tags)


def reward_score_accuracy(completion: str, gt_final: int) -> float:
    """
    Smooth reward based on absolute error between predicted and GT final score.
    Uses a Gaussian-like decay: reward = exp(-0.5 * err²)
    Perfect match → 1.0, off-by-1 → 0.61, off-by-2 → 0.14, ≥4 → ~0.0
    """
    pred = extract_multi_metrics(completion)["final"]
    err  = abs(pred - gt_final)
    return float(math.exp(-0.5 * err ** 2))


def reward_metric_accuracy(completion: str, gt_sf: int, gt_cp: int, gt_rq: int) -> float:
    """
    Average Gaussian reward across the 3 sub-metrics.
    Penalizes missing metrics by giving 0.0 for each absent tag.
    """
    parsed = extract_multi_metrics(completion)

    def _gauss(pred, gt):
        return math.exp(-0.5 * (pred - gt) ** 2)

    r_sf = _gauss(parsed["sf"], gt_sf) if parsed["sf_found"] else 0.0
    r_cp = _gauss(parsed["cp"], gt_cp) if parsed["cp_found"] else 0.0
    r_rq = _gauss(parsed["rq"], gt_rq) if parsed["rq_found"] else 0.0

    return (r_sf + r_cp + r_rq) / 3.0


def reward_reasoning(completion: str) -> float:
    """
    Scores reasoning chain quality:
      - Presence of 3-step analysis (Style Extraction, Analogy Check, Conflict Check)
      - Min 80 characters inside <Thinking_Process>
      - No degenerate "..." or repetitive output
    """
    score = 0.0
    thinking_match = re.search(
        r"<Thinking_Process>(.*?)</Thinking_Process>", completion, re.DOTALL | re.I
    )
    if not thinking_match:
        return 0.0

    thinking_text = thinking_match.group(1).strip()
    if len(thinking_text) < 80:
        return 0.1   # present but nearly empty

    # Check 3-step keywords
    step_hits = sum([
        bool(re.search(r"Step 1|Style Extraction",  thinking_text, re.I)),
        bool(re.search(r"Step 2|Analogy Check",      thinking_text, re.I)),
        bool(re.search(r"Step 3|Conflict Check",     thinking_text, re.I)),
    ])
    score += step_hits * 0.25          # 0.75 max from steps

    # Length bonus (soft cap at 300 chars)
    len_bonus = min(len(thinking_text) / 300.0, 1.0) * 0.25
    score += len_bonus

    # Degenerate output penalty
    if "..." in thinking_text or thinking_text.count("\n") < 2:
        score *= 0.5

    return min(score, 1.0)


def compute_composite_reward(
    completion: str,
    gt_final: int, gt_sf: int, gt_cp: int, gt_rq: int
) -> float:
    """
    Weighted composite of all 4 reward signals → clipped to [0, 1].
    """
    r_fmt  = reward_format(completion)
    r_sacc = reward_score_accuracy(completion, gt_final)
    r_macc = reward_metric_accuracy(completion, gt_sf, gt_cp, gt_rq)
    r_reas = reward_reasoning(completion)

    composite = (
        W_FORMAT    * r_fmt  +
        W_SCORE_ACC * r_sacc +
        W_METRIC_ACC * r_macc +
        W_REASONING * r_reas
    )
    return float(np.clip(composite, 0.0, 1.0))


# ── Sanity-check reward functions ─────────────────────────────────────────────
_test_output = """<Thinking_Process>
Step 1: Style Extraction: Bold halftone dots, flat primary colours, thick black outlines.
Step 2: Analogy Check: Query result shows clear dot patterns and flat colour fill.
Step 3: Conflict Check: Minor edge artefacts but no content corruption.
</Thinking_Process>
<Quantitative_Metrics>
Style Fidelity: 8
Content Preservation: 9
Rendering Quality: 8
</Quantitative_Metrics>
<Final_Output>
Brief Rationale: Strong pop_art transfer with well-realised halftone patterns.
</Final_Output>"""

_r = compute_composite_reward(_test_output, gt_final=8, gt_sf=8, gt_cp=9, gt_rq=8)
print(f"[REWARD] Sanity-check on a near-perfect output → composite reward = {_r:.4f}  (expect ≥ 0.85)")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 17 ── GRPO DATASET  (Wraps the same CSVs from SFT, prompt-only)
# ═══════════════════════════════════════════════════════════════════════════════

# Reuse SYSTEM_PROMPT, build_user_content, STYLE_DESCRIPTIONS from SFT cells
# (They remain in the kernel namespace)

class GRPOStyleDataset(Dataset):
    """
    Returns PROMPT-ONLY messages (no assistant turn).
    Ground-truth scores are returned as metadata for the reward function.
    Samples are shuffled each epoch for GRPO diversity.
    """
    def __init__(self, dataframe: pd.DataFrame, shuffle: bool = True):
        self.samples = dataframe.to_dict("records")
        if shuffle:
            random.shuffle(self.samples)
        print(f"  [GRPO DATASET] {len(self.samples)} prompts ready.")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        s = self.samples[idx]
        style      = s["style"]
        style_desc = STYLE_DESCRIPTIONS.get(style, "distinctive textures, colors, and brushstrokes.")

        # Load images lazily
        def _img(p):
            return Image.open(p).convert("RGB").resize((IMAGE_SIZE, IMAGE_SIZE), Image.LANCZOS)

        user_content = build_user_content(
            _img(s["ex_a_src_path"]), _img(s["ex_a_dst_path"]),
            _img(s["ex_b_src_path"]), _img(s["ex_b_dst_path"]),
            _img(s["q_src_path"]),    _img(s["q_res_path"]),
            style, style_desc,
        )

        messages = [
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user",   "content": user_content},
        ]

        return {
            "messages":    messages,
            "gt_final":    int(s["final_score"]),
            "gt_sf":       int(s.get("sf_score",  s["final_score"])),  # fallback if not present
            "gt_cp":       int(s.get("cp_score",  s["final_score"])),
            "gt_rq":       int(s.get("rq_score",  s["final_score"])),
            "style":       style,
            "morph_idx":   int(s["morph_idx"]),
        }


print("[GRPO] Building datasets …")
grpo_train_ds = GRPOStyleDataset(df_train, shuffle=True)

# Use a small eval subset for speed (GRPO eval is expensive — generation required)
grpo_val_ds   = GRPOStyleDataset(df_val.sample(n=min(200, len(df_val)), random_state=SEED), shuffle=False)

print(f"  GRPO train prompts : {len(grpo_train_ds)}")
print(f"  GRPO val prompts   : {len(grpo_val_ds)}")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 18 ── CORE GRPO ENGINE  (Custom VLM-aware implementation)
#
# Why custom instead of trl.GRPOTrainer?
#   → trl.GRPOTrainer targets text-only models (as of trl ≤ 0.10).
#   → Unsloth's FastVisionModel requires process_vision_info() for image tokens.
#   → We need full control over group sampling, reward shaping, and KL loss.
# ═══════════════════════════════════════════════════════════════════════════════

class GRPOVisionEngine:
    """
    Group Relative Policy Optimization for Unsloth Vision-Language Models.

    Algorithm (per mini-batch of B prompts):
      1.  For each prompt i, generate G completions (no grad, policy model).
      2.  Score each completion with composite_reward().
      3.  Compute group-relative advantages:
              A_{i,g} = (r_{i,g} - mean_g(r_i)) / (std_g(r_i) + eps)
      4.  Re-run forward pass on all B×G completions WITH grad.
      5.  Compute clipped policy gradient loss (PPO-style clip).
      6.  Add KL divergence penalty vs frozen reference model.
      7.  Backpropagate & step.
    """

    def __init__(
        self,
        policy_model,
        ref_model,
        tokenizer,
        optimizer,
        scheduler,
        group_size:   int   = GRPO_GROUP_SIZE,
        kl_coef:      float = GRPO_KL_COEF,
        clip_eps:     float = GRPO_CLIP_EPS,
        max_new_tokens: int = GRPO_MAX_NEW_TOKENS,
        device: str   = "cuda",
    ):
        self.policy    = policy_model
        self.ref       = ref_model
        self.tok       = tokenizer
        self.opt       = optimizer
        self.sched     = scheduler
        self.G         = group_size
        self.kl_coef   = kl_coef
        self.clip_eps  = clip_eps
        self.max_new   = max_new_tokens
        self.device    = device
        self.step_logs: List[Dict] = []

    # ── Tokenise a single message dict into model inputs ──────────────────────
    def _prepare_inputs(self, messages: List[Dict]) -> Dict:
        text = self.tok.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=True
        )
        image_inputs, _ = process_vision_info(messages)
        inputs = self.tok(
            text=[text], images=image_inputs,
            padding=True, return_tensors="pt"
        )
        return {k: v.to(self.device) for k, v in inputs.items()}

    # ── Generate G completions for ONE prompt ─────────────────────────────────
    @torch.inference_mode()
    def _sample_group(self, messages: List[Dict]) -> List[str]:
        inputs       = self._prepare_inputs(messages)
        prompt_len   = inputs["input_ids"].shape[1]

        out_ids = self.policy.generate(
            **inputs,
            max_new_tokens = self.max_new,
            do_sample      = True,
            temperature    = 0.8,      # diversity within group
            top_p          = 0.92,
            num_return_sequences = self.G,
            eos_token_id   = self.tok.eos_token_id,
            pad_token_id   = self.tok.pad_token_id,
        )

        completions = self.tok.batch_decode(
            out_ids[:, prompt_len:], skip_special_tokens=True
        )
        return completions   # list of G strings

    # ── Per-token log-probs for a (prompt+completion) pair ────────────────────
    def _log_probs(self, model, messages: List[Dict], completion: str) -> torch.Tensor:
        """
        Returns a 1-D tensor of per-token log-probabilities for the completion
        tokens only (excluding the prompt).
        """
        # Build full conversation (prompt + assistant response)
        full_msgs = messages + [{"role": "assistant", "content": completion}]
        text      = self.tok.apply_chat_template(full_msgs, tokenize=False)
        image_inputs, _ = process_vision_info(messages)

        enc = self.tok(
            text=[text], images=image_inputs,
            padding=True, return_tensors="pt"
        )
        enc = {k: v.to(self.device) for k, v in enc.items()}

        # Prompt-only length for masking
        prompt_text = self.tok.apply_chat_template(
            messages, tokenize=False, add_generation_prompt=True
        )
        prompt_enc  = self.tok(text=[prompt_text], return_tensors="pt")
        prompt_len  = prompt_enc["input_ids"].shape[1]

        with torch.cuda.amp.autocast(dtype=torch.bfloat16):
            out    = model(**enc)
            logits = out.logits                              # [1, L, V]

        # Shift to align logits → labels
        shift_logits = logits[0, :-1, :]                    # [L-1, V]
        shift_ids    = enc["input_ids"][0, 1:]              # [L-1]

        log_probs_all = torch.log_softmax(shift_logits, dim=-1)  # [L-1, V]
        token_log_probs = log_probs_all.gather(
            1, shift_ids.unsqueeze(1)
        ).squeeze(1)                                         # [L-1]

        # Keep only completion tokens
        completion_start = max(0, prompt_len - 1)
        completion_lp    = token_log_probs[completion_start:]

        return completion_lp   # [comp_len]

    # ── KL divergence: policy vs reference (token-level mean) ─────────────────
    def _kl_divergence(self, messages, completion) -> torch.Tensor:
        """
        KL(π_θ ‖ π_ref) approximated per token as log π_θ - log π_ref.
        """
        lp_policy = self._log_probs(self.policy, messages, completion)

        with torch.inference_mode():
            lp_ref   = self._log_probs(self.ref,    messages, completion)

        # Align lengths (generation may differ slightly due to padding)
        min_len = min(lp_policy.shape[0], lp_ref.shape[0])
        kl = (lp_policy[:min_len] - lp_ref[:min_len].detach()).mean()
        return kl

    # ── Single GRPO step over one mini-batch of prompts ───────────────────────
    def grpo_step(self, batch: List[Dict]) -> Dict:
        """
        batch: list of dataset items (messages + gt labels).
        Returns a dict of scalar metrics for logging.
        """
        all_pg_losses, all_kl_losses, all_rewards = [], [], []

        for item in batch:
            messages  = item["messages"]
            gt_final  = item["gt_final"]
            gt_sf     = item["gt_sf"]
            gt_cp     = item["gt_cp"]
            gt_rq     = item["gt_rq"]

            # ── Step 1: Sample G completions (no grad) ────────────────────────
            FastVisionModel.for_inference(self.policy)
            completions = self._sample_group(messages)
            FastVisionModel.for_training(self.policy)

            # ── Step 2: Score each completion ─────────────────────────────────
            rewards = [
                compute_composite_reward(c, gt_final, gt_sf, gt_cp, gt_rq)
                for c in completions
            ]

            rewards_t = torch.tensor(rewards, dtype=torch.float32, device=self.device)
            mean_r    = rewards_t.mean()
            std_r     = rewards_t.std().clamp(min=1e-8)

            # ── Step 3: Group-relative advantages ─────────────────────────────
            advantages = (rewards_t - mean_r) / std_r   # shape [G]

            # ── Step 4 & 5: Policy gradient with clipping ─────────────────────
            pg_loss_group = torch.tensor(0.0, device=self.device, requires_grad=True)
            kl_loss_group = torch.tensor(0.0, device=self.device, requires_grad=True)

            for g_idx, (comp, adv) in enumerate(zip(completions, advantages)):
                # Forward through policy WITH grad
                lp_pol = self._log_probs(self.policy, messages, comp)

                # Forward through reference (no grad — already frozen)
                with torch.inference_mode():
                    lp_ref = self._log_probs(self.ref, messages, comp)

                min_len  = min(lp_pol.shape[0], lp_ref.shape[0])
                lp_pol_a = lp_pol[:min_len]
                lp_ref_a = lp_ref[:min_len].detach()

                # Per-sequence log prob ratio
                log_ratio  = (lp_pol_a - lp_ref_a).sum()         # scalar
                ratio      = torch.exp(log_ratio.clamp(-20, 20))  # numerical safety

                # PPO-style clipped objective (negative because we *minimise* loss)
                pg_clip = torch.min(
                    ratio      * adv,
                    ratio.clamp(1 - self.clip_eps, 1 + self.clip_eps) * adv,
                )
                pg_loss_group = pg_loss_group + (-pg_clip)

                # KL penalty (using per-token KL)
                kl = (lp_pol_a - lp_ref_a).mean()                 # ≈ KL per token
                kl_loss_group = kl_loss_group + kl

            pg_loss_group  = pg_loss_group  / self.G
            kl_loss_group  = kl_loss_group  / self.G
            total_item_loss = pg_loss_group + self.kl_coef * kl_loss_group

            all_pg_losses.append(pg_loss_group)
            all_kl_losses.append(kl_loss_group)
            all_rewards.extend(rewards)

        # ── Step 6: Aggregate & backprop over mini-batch ──────────────────────
        total_loss = torch.stack(all_pg_losses).mean() + \
                     self.kl_coef * torch.stack(all_kl_losses).mean()

        total_loss.backward()

        return {
            "loss":     total_loss.item(),
            "pg_loss":  torch.stack(all_pg_losses).mean().item(),
            "kl_loss":  torch.stack(all_kl_losses).mean().item(),
            "mean_reward": float(np.mean(all_rewards)),
            "max_reward":  float(np.max(all_rewards)),
            "min_reward":  float(np.min(all_rewards)),
        }


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 19 ── OPTIMIZER, SCHEDULER, LOGGING UTILS
# ═══════════════════════════════════════════════════════════════════════════════
import csv as _csv

# Only update LoRA adapters (already set from SFT)
_trainable_params = [p for p in model.parameters() if p.requires_grad]
print(f"[GRPO] Trainable parameter tensors: {len(_trainable_params)}")

optimizer = torch.optim.AdamW(
    _trainable_params,
    lr           = GRPO_LR,
    betas        = (0.9, 0.999),
    eps          = 1e-8,
    weight_decay = 0.01,
)

# Cosine decay with linear warm-up
_total_grpo_steps = (len(grpo_train_ds) // (GRPO_MINI_BATCH * GRPO_GRAD_ACCUM)) * GRPO_EPOCHS

def _lr_lambda(step):
    if step < GRPO_WARMUP_STEPS:
        return step / max(1, GRPO_WARMUP_STEPS)
    progress = (step - GRPO_WARMUP_STEPS) / max(1, _total_grpo_steps - GRPO_WARMUP_STEPS)
    return 0.5 * (1.0 + math.cos(math.pi * progress))

scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, _lr_lambda)


class GRPOLogger:
    def __init__(self, log_csv: str):
        self.log_csv = log_csv
        self._header_written = False

    def log(self, row: Dict):
        # Print to stdout
        print(
            f"[GRPO] step={row['step']:>5} | ep={row['epoch']:.3f} | "
            f"loss={row['loss']:>7.4f} | pg={row['pg_loss']:>7.4f} | "
            f"kl={row['kl_loss']:>6.4f} | "
            f"r̄={row['mean_reward']:.4f} | lr={row['lr']:.2e}"
        )
        # Write to CSV
        write_header = not os.path.exists(self.log_csv) or not self._header_written
        with open(self.log_csv, "a", newline="") as f:
            writer = _csv.DictWriter(f, fieldnames=list(row.keys()))
            if write_header:
                writer.writeheader()
                self._header_written = True
            writer.writerow(row)

    def log_eval(self, step: int, mean_reward: float, best: float) -> bool:
        is_best = mean_reward > best
        marker  = "✅ NEW BEST" if is_best else "(no improvement)"
        print(f"\n[GRPO EVAL] step={step} | mean_reward={mean_reward:.4f} | "
              f"best={best:.4f} | {marker}\n")
        return is_best


logger   = GRPOLogger(GRPO_LOG_CSV)
engine   = GRPOVisionEngine(
    policy_model  = model,
    ref_model     = ref_model,
    tokenizer     = tokenizer,
    optimizer     = optimizer,
    scheduler     = scheduler,
)

print(f"[GRPO] Engine ready. Total optimizer steps: {_total_grpo_steps:,}")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 20 ── GRPO EVAL LOOP  (runs on grpo_val_ds subset)
# ═══════════════════════════════════════════════════════════════════════════════
@torch.inference_mode()
def run_grpo_eval(num_samples: int = 40) -> float:
    """
    Runs greedy decoding on num_samples val prompts, computes composite rewards.
    Returns mean composite reward as the tracked eval metric.
    """
    FastVisionModel.for_inference(model)
    rewards = []

    for i in range(min(num_samples, len(grpo_val_ds))):
        item      = grpo_val_ds[i]
        messages  = item["messages"]
        gt_final  = item["gt_final"]
        gt_sf     = item["gt_sf"]
        gt_cp     = item["gt_cp"]
        gt_rq     = item["gt_rq"]

        text         = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
        image_inputs, _ = process_vision_info(messages)
        inputs       = tokenizer(
            text=[text], images=image_inputs, padding=True, return_tensors="pt"
        ).to("cuda")

        out_ids = model.generate(
            **inputs,
            max_new_tokens  = GRPO_MAX_NEW_TOKENS,
            do_sample       = False,
            eos_token_id    = tokenizer.eos_token_id,
            pad_token_id    = tokenizer.pad_token_id,
        )
        completion = tokenizer.batch_decode(
            out_ids[:, inputs["input_ids"].shape[1]:], skip_special_tokens=True
        )[0]

        r = compute_composite_reward(completion, gt_final, gt_sf, gt_cp, gt_rq)
        rewards.append(r)

        del inputs, out_ids, image_inputs
        flush()

    FastVisionModel.for_training(model)
    return float(np.mean(rewards))


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 21 ── EXECUTE GRPO TRAINING LOOP
# ═══════════════════════════════════════════════════════════════════════════════
from torch.utils.data import DataLoader

def _collate_identity(batch):
    """No collation — each item is a dict returned as-is."""
    return batch

grpo_loader = DataLoader(
    grpo_train_ds,
    batch_size  = GRPO_MINI_BATCH,
    shuffle     = True,
    collate_fn  = _collate_identity,
    num_workers = 0,   # images in main process (PIL not fork-safe)
)

Path(GRPO_OUTPUT_DIR).mkdir(parents=True, exist_ok=True)

global_step   = 0
best_reward   = -float("inf")
grad_steps    = 0
t_start       = time.time()
MAX_TRAIN_SEC = 11.0 * 3600   # safety limit: 11 hours

print(f"\n{'═'*70}")
print(f" GRPO Training — {GRPO_EPOCHS} epoch(s) | {_total_grpo_steps:,} optimizer steps")
print(f"{'═'*70}\n")

FastVisionModel.for_training(model)
optimizer.zero_grad()

for epoch in range(GRPO_EPOCHS):
    for batch in grpo_loader:
        # ── Time-limit safety check ───────────────────────────────────────────
        if time.time() - t_start > MAX_TRAIN_SEC:
            print("[TIMEOUT] Reached 11 h limit — saving and stopping.")
            break

        # ── GRPO step ─────────────────────────────────────────────────────────
        metrics = engine.grpo_step(batch)
        grad_steps += 1

        # ── Gradient accumulation ─────────────────────────────────────────────
        if grad_steps % GRPO_GRAD_ACCUM == 0:
            torch.nn.utils.clip_grad_norm_(_trainable_params, max_norm=1.0)
            optimizer.step()
            scheduler.step()
            optimizer.zero_grad()
            global_step += 1

            # ── Logging ───────────────────────────────────────────────────────
            if global_step % GRPO_LOG_STEPS == 0:
                row = {
                    "timestamp":   datetime.now().isoformat(timespec="seconds"),
                    "epoch":       round(epoch + grad_steps / len(grpo_loader), 3),
                    "step":        global_step,
                    "loss":        metrics["loss"],
                    "pg_loss":     metrics["pg_loss"],
                    "kl_loss":     metrics["kl_loss"],
                    "mean_reward": metrics["mean_reward"],
                    "max_reward":  metrics["max_reward"],
                    "min_reward":  metrics["min_reward"],
                    "lr":          optimizer.param_groups[0]["lr"],
                }
                logger.log(row)

            # ── Evaluation & best-checkpoint saving ───────────────────────────
            if global_step % GRPO_EVAL_STEPS == 0:
                eval_reward = run_grpo_eval(num_samples=40)
                is_best     = logger.log_eval(global_step, eval_reward, best_reward)

                if is_best:
                    best_reward = eval_reward
                    best_path   = Path(GRPO_OUTPUT_DIR) / "best"
                    model.save_pretrained(str(best_path))
                    tokenizer.save_pretrained(str(best_path))
                    print(f"  💾 Best GRPO adapter saved → {best_path}")

            flush()

    else:
        continue  # inner loop completed; continue outer
    break         # inner loop was broken (timeout); stop outer too

_elapsed = (time.time() - t_start) / 60
print(f"\n[GRPO] Training complete in {_elapsed:.1f} min")
print(f"[GRPO] Best eval reward : {best_reward:.4f}")


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 22 ── SAVE FINAL GRPO ADAPTER + OPTIONAL MERGED MODEL
# ═══════════════════════════════════════════════════════════════════════════════

# Always save the final state regardless of best-checkpoint
final_path = Path(GRPO_OUTPUT_DIR) / "final"
model.save_pretrained(str(final_path))
tokenizer.save_pretrained(str(final_path))
print(f"[GRPO] Final adapter saved → {final_path}")

# ── Optional: merge LoRA into base weights and save as a stand-alone model ───
MERGE_AND_SAVE = True   # set False to skip (saves ~20 min)

if MERGE_AND_SAVE:
    print("\n[MERGE] Merging LoRA adapter into base weights …")
    flush()

    merged = model.merge_and_unload()   # returns the base model with merged weights
    merged_path = Path(GRPO_OUTPUT_DIR) / "merged"
    merged.save_pretrained(str(merged_path), safe_serialization=True)
    tokenizer.save_pretrained(str(merged_path))

    print(f"[MERGE] Merged model saved → {merged_path}")
    del merged
    flush()


# ═══════════════════════════════════════════════════════════════════════════════
# CELL 23 ── GRPO POST-TRAINING EVALUATION  (50 val samples, detailed)
# ═══════════════════════════════════════════════════════════════════════════════
print("\n[EVAL] Loading best GRPO adapter for final evaluation …")
FastVisionModel.for_inference(model)
model.load_adapter(str(Path(GRPO_OUTPUT_DIR) / "best"), adapter_name="default")

grpo_val_preds = []

for i in range(min(50, len(grpo_val_ds))):
    item     = grpo_val_ds[i]
    messages = item["messages"]

    text         = tokenizer.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    image_inputs, _ = process_vision_info(messages)
    inputs       = tokenizer(
        text=[text], images=image_inputs, padding=True, return_tensors="pt"
    ).to("cuda")

    with torch.inference_mode():
        out_ids = model.generate(
            **inputs,
            max_new_tokens = GRPO_MAX_NEW_TOKENS,
            do_sample      = False,
            eos_token_id   = tokenizer.eos_token_id,
            pad_token_id   = tokenizer.pad_token_id,
        )

    completion = tokenizer.batch_decode(
        out_ids[:, inputs["input_ids"].shape[1]:], skip_special_tokens=True
    )[0]

    parsed     = extract_multi_metrics(completion)
    gt_final   = item["gt_final"]
    err        = abs(parsed["final"] - gt_final)
    composite  = compute_composite_reward(
        completion, gt_final, item["gt_sf"], item["gt_cp"], item["gt_rq"]
    )

    grpo_val_preds.append({
        "sample_idx":       i,
        "style":            item["style"],
        "morph_idx":        item["morph_idx"],
        "gt_final":         gt_final,
        "pred_final":       parsed["final"],
        "pred_sf":          parsed["sf"],
        "pred_cp":          parsed["cp"],
        "pred_rq":          parsed["rq"],
        "abs_error":        err,
        "exact_match":      (err == 0),
        "within_1":         (err <= 1),
        "composite_reward": composite,
        "format_reward":    reward_format(completion),
        "reasoning_reward": reward_reasoning(completion),
        "output_length":    len(completion),
        "raw_output":       completion,
    })

    del inputs, out_ids, image_inputs
    flush()

    if (i + 1) % 10 == 0:
        cur = grpo_val_preds[-(i+1):]
        m   = np.mean([p["abs_error"] for p in grpo_val_preds])
        acc = sum(p["exact_match"] for p in grpo_val_preds) / len(grpo_val_preds)
        print(f"  [{i+1:>3}/50] MAE so far: {m:.3f} | Exact Acc: {acc*100:.1f}%")

# ── Summary ───────────────────────────────────────────────────────────────────
if grpo_val_preds:
    errors   = [p["abs_error"]        for p in grpo_val_preds]
    rewards  = [p["composite_reward"] for p in grpo_val_preds]
    n        = len(grpo_val_preds)

    print("\n" + "═"*70)
    print(" GRPO Final Val Summary (50 samples)")
    print("═"*70)
    print(f"  Score MAE              : {np.mean(errors):.3f}")
    print(f"  Exact match (±0)       : {sum(p['exact_match'] for p in grpo_val_preds)}/{n} "
          f"({100*sum(p['exact_match'] for p in grpo_val_preds)/n:.1f}%)")
    print(f"  Within ±1              : {sum(p['within_1'] for p in grpo_val_preds)}/{n} "
          f"({100*sum(p['within_1'] for p in grpo_val_preds)/n:.1f}%)")
    print(f"  Mean composite reward  : {np.mean(rewards):.4f}")
    print(f"  Mean format reward     : {np.mean([p['format_reward'] for p in grpo_val_preds]):.4f}")
    print(f"  Mean reasoning reward  : {np.mean([p['reasoning_reward'] for p in grpo_val_preds]):.4f}")
    print(f"  Avg output length      : {np.mean([p['output_length'] for p in grpo_val_preds]):.0f} chars")
    print("═"*70)

    # Save predictions
    with open("/kaggle/working/grpo_val_predictions.json", "w") as f:
        json.dump(grpo_val_preds, f, indent=2)
    print("\n[EVAL] Predictions saved → /kaggle/working/grpo_val_predictions.json")
```
